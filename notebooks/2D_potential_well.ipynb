```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def add_gaussian_noise(data, mean=0, std=1):
  """Adds Gaussian noise to the input data.

  Args:
    data: A NumPy array or a list of numerical values.
    mean: The mean of the Gaussian distribution (default: 0).
    std: The standard deviation of the Gaussian distribution (default: 1).

  Returns:
    A NumPy array with added Gaussian noise.
  """
  noise = np.random.normal(mean, std, size=data.shape)
  noisy_data = data + noise
  return noisy_data
```

```python
# Grid parameters
N = 1000  # number of points along each axis
L = 3 * np.pi  # physical length of the grid

# Create grid
x = np.linspace(0, L, N)
y = np.linspace(0, L, N)
X, Y = np.meshgrid(x, y)

# "Egg carton" potential: periodic sinusoidal energy landscape
Z = (np.sin(X) * np.sin(Y)+1)/2
Z2 = (np.sin(X+np.pi/6) * np.sin(Y+np.pi/6)+1)/2

# 3D Plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z+Z2, cmap='jet', edgecolor='none')

# Plot styling
ax.set_title('Egg Carton Energy Landscape')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('Energy')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()
```

```python
Z_noisy = add_gaussian_noise(Z,mean=0,std=0.1)
```

```python
# 3D Plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(X, Y, Z_noisy, cmap='jet', edgecolor='none')

# Plot styling
ax.set_title('Egg Carton Energy Landscape')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('Energy')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()
```
