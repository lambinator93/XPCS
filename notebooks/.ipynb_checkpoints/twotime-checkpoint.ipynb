```python
CONFIG_FILE_NAME = "NPS_11092023_processed_config.json"

figure_path = '/Users/eriklamb/Documents/Publications/NPS_Petra/Figures/'


# Define the path to the configuration file
config_path = Path("/Users/eriklamb/Documents/Python/Projects/XPCS/configs/"+CONFIG_FILE_NAME)

# Open and load the JSON configuration file
with open(config_path, "r") as file:
    config = json.load(file)
    #json_str = json.loads(jsonString, strict=False)
```

```python
T = 298

BASE = Path(config.get("BASE"))
file_processed = BASE/Path(config.get(f"{T}_processed"))
file_results = BASE/Path(config.get(f"{T}_results"))

with h5py.File(file_processed, 'r') as DATA:
    det          = DATA['data/det'][:]
    frameSpacing = DATA['experimental_parameters/frame_spacing'][:][0]
    
with h5py.File(file_results, 'r') as DATA:
    masks        = DATA['XPCS/Mask'][:]
    dQ           = DATA['XPCS/dQ'][:]
```

```python
fig, ax = plt.subplots(2,5,figsize=(25,25))

m = 0
for i in range(2):
    for j in range(5):
        IMG_temp = det[:,masks[m,:,:]==1]
        #IMG_temp = det[:,mask_sum>0]
        IMGS_NORM = IMG_temp.reshape((IMG_temp.shape[0],IMG_temp.shape[1],1))
        #IMGS_NORM = IMG_temp.reshape((650,IMG_temp.shape[1],1))


        C = twotime(IMGS_NORM)#/np.size(det[:,masks[M,:,:]==1])
        C_extent = [0,frameSpacing*det.shape[0],0,frameSpacing*det.shape[0]]

        C_norm = (C-np.min(C))/(np.max(C)-np.min(C))
        
        
        ax[i,j].imshow(C_norm,origin="lower",extent=C_extent,vmin = 0,vmax = 1cmap='plasma')
        ax[i,j].set_ylabel('t2 (s)')
        ax[i,j].set_xlabel('t1 (s)')
        ax[i,j].set_title(f'dQ = {dQ[m]} 1/Å')
        
        
        axs[i,j].xaxis.set_major_formatter(ticker.ScalarFormatter(useMathText=True))
        axs[i,j].xaxis.get_offset_text().set_fontsize(12)  # Adjust font size if needed
        axs[i,j].xaxis.get_offset_text().set_position((1.02, 0))  # Move offset label slightly

        axs[i,j].yaxis.set_major_formatter(ticker.ScalarFormatter(useMathText=True))
        axs[i,j].yaxis.get_offset_text().set_fontsize(12)
        axs[i,j].yaxis.get_offset_text().set_position((0, 1.02))  # Move offset label slightly

        axs[i,j].ticklabel_format(axis='both', style='sci', scilimits=(0, 0))  # Force scientific notation

        axs[i,j].tick_params(axis='both', which='major', labelsize=12)
        axs[i,j].tick_params(axis='both', which='minor', labelsize=12)


        m+=1
#fig.colorbar(im, ax=ax)

plt.show
```
