```python
import numpy as np

def compute_g2(I_stack, max_tau=10, dq_list=[(1, 1)]):
    """
    Compute g2(q0, Δq, τ) for a given stack of intensity images.
    
    Parameters:
        I_stack : ndarray
            3D array of shape (T, Nx, Ny) — intensity images over time
        max_tau : int
            Maximum τ to compute
        dq_list : list of tuples
            List of (Δqx, Δqy) shift vectors

    Returns:
        g2_dict : dict
            Dictionary keyed by (Δqx, Δqy) with values being arrays of g2(τ)
    """
    T, Nx, Ny = I_stack.shape
    mean_I = np.mean(I_stack)
    
    g2_dict = {}

    for dq in dq_list:
        dqx, dqy = dq
        g2_tau = []

        for tau in range(1, max_tau + 1):
            corr_sum = 0
            count = 0

            for t in range(T - tau):
                I1 = I_stack[t]
                I2 = np.roll(I_stack[t + tau], shift=(-dqy, -dqx), axis=(0, 1))

                # Exclude wrapped-around edges (simple mask)
                valid_slice_x = slice(max(0, dqx), min(Nx, Nx + dqx))
                valid_slice_y = slice(max(0, dqy), min(Ny, Ny + dqy))

                I1_valid = I1[valid_slice_x, valid_slice_y]
                I2_valid = I2[valid_slice_x, valid_slice_y]

                corr_sum += np.mean(I1_valid * I2_valid)
                count += 1

            g2_value = corr_sum / count / (mean_I ** 2)
            g2_tau.append(g2_value)

        g2_dict[dq] = np.array(g2_tau)

    return g2_dict
```
